```python
import pandas as pd
```

```python
df = pd.read_csv('data/screening_log.csv')
```

```python
is_record_id_unique = not df['record_id'].duplicated().any()
```

```python
if is_record_id_unique:
    print("✅ Harika! 'record_id' sütunundaki tüm değerler benzersiz (unique).\n")
else:
    print("❌ Dikkat! 'record_id' sütununda yinelenen değerler bulundu.")
    # Yinelenen 'record_id' değerlerinin sayısını bulalım
    num_duplicates = df['record_id'].duplicated().sum()
    print(f"   Toplam {num_duplicates} adet yinelenen 'record_id' kaydı var.")
    
    # Hangi 'record_id'lerin tekrar ettiğini ve kaçar kez tekrar ettiğini görelim
    duplicated_ids = df['record_id'].value_counts()
    print("   Yinelenen ID'ler ve tekrar sayıları:")
    display(duplicated_ids[duplicated_ids > 1])
    print("-" * 40)
```
