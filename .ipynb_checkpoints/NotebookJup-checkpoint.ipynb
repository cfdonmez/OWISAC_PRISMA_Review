```python
import pandas as pd
import os

# Dosya yolunu düzelt
file_path = 'data/included_studies.csv'
if not os.path.exists(file_path):
    print(f"❌ Dosya bulunamadı: {file_path}")
    print(f"Mevcut dizin: {os.getcwd()}")
    print(f"Dizindeki dosyalar: {os.listdir('.')}")
else:
    # Dosyayı oku
    df_included = pd.read_csv(file_path)
    
    # DataFrame bilgilerini göster
    print(f"Toplam kayıt sayısı: {len(df_included)}")
    print(f"\nSütunlar: {df_included.columns.tolist()}")
    print(f"\nİlk 5 satır:")
    print(df_included.head())
    
    # record_id duplicate kontrolü
    print("\n" + "="*80)
    print("RECORD_ID DUPLICATE KONTROLÜ")
    print("="*80)
    record_id_dups = df_included[df_included['record_id'].duplicated(keep=False)]
    if len(record_id_dups) > 0:
        print(f"❌ {len(record_id_dups)} kayıtta record_id tekrarı bulundu!")
        print(f"Tekrar eden benzersiz record_id sayısı: {record_id_dups['record_id'].nunique()}")
        print("\nTekrar eden record_id'ler:")
        print(record_id_dups[['record_id', 'title']].sort_values('record_id'))
    else:
        print("✅ record_id sütununda duplicate YOK")
    
    # title duplicate kontrolü
    print("\n" + "="*80)
    print("TITLE DUPLICATE KONTROLÜ")
    print("="*80)
    title_dups = df_included[df_included['title'].duplicated(keep=False)]
    if len(title_dups) > 0:
        print(f"❌ {len(title_dups)} kayıtta title tekrarı bulundu!")
        print(f"Tekrar eden benzersiz title sayısı: {title_dups['title'].nunique()}")
        print("\nTekrar eden title'lar ve record_id'leri:")
        print(title_dups[['record_id', 'title']].sort_values('title'))
    else:
        print("✅ title sütununda duplicate YOK")
    
    # Özet
    print("\n" + "="*80)
    print("ÖZET")
    print("="*80)
    print(f"Toplam kayıt: {len(df_included)}")
    print(f"record_id duplicate: {len(record_id_dups)} kayıt")
    print(f"title duplicate: {len(title_dups)} kayıt")
```

```python
# Temizlenmiş screening veri setindeki tüm benzersiz record_id'leri al
# Bu, screening.py dosyanızdaki son işlemden sonraki screening dataframe'i olmalı
cleaned_screening_ids = screening['record_id'].unique()

# included veri setini, sadece cleaned_screening_ids içinde olan record_id'lere göre filtrele
included_updated = included[included['record_id'].isin(cleaned_screening_ids)].copy()

print(f"included veri seti, screening'deki benzersiz record_id'lerle güncellendi.")
print(f"Yeni included veri setinde {len(included_updated)} kayıt bulunmaktadır.")

# Güncellenmiş included veri setini bir değişkene atayın veya bir dosyaya kaydedin
# Örneğin, included = included_updated diyerek bir sonraki adımlara geçebilirsiniz
# included_updated.to_csv('data/included_studies_updated.csv', index=False)
```

```python
import pandas as pd
from IPython.display import display

print("=== AYNI RECORD_ID, FARKLI TITLE KONTROLÜ ===")

# included_updated veri setini 'record_id'ye göre grupluyoruz.
# Ardından, 'title' sütunundaki benzersiz eleman sayısının 1'den fazla olduğu grupları filtreliyoruz.
same_id_diff_title = included_updated.groupby('record_id').filter(lambda x: len(x['title'].unique()) > 1)

# Bulunan kayıt sayısını ekrana yazdırıyoruz.
print(f"Toplam {len(same_id_diff_title)} kayıt bulundu.")

# Problemli kayıtların ham verisini gösteriyoruz.
display(same_id_diff_title)

print("\n" + "="*80 + "\n")

# --- Detaylı Analiz ---
# Bulduğumuz problemli kayıtları daha anlaşılır bir tabloya dönüştürüyoruz.
print("=== DETAYLI ANALİZ: FARKLI BAŞLIKLAR ===")

grouped_analysis = same_id_diff_title.groupby('record_id').agg(
    titles=('title', list),         # Aynı ID'ye ait tüm başlıkları liste olarak topluyoruz.
    authors=('authors', list),      # Aynı ID'ye ait tüm yazarları liste olarak topluyoruz.
    record_count=('record_id', 'count') # Her ID'nin kaç kez tekrar ettiğini sayıyoruz.
).reset_index()

# Analiz tablosunu ekrana basıyoruz.
display(grouped_analysis)
```

```python
import pandas as pd
from IPython.display import display

# --- 1. Adım: Aynı record_id'ye sahip, farklı başlıklı kayıtları bulma ---
# Bu kod, veri setinizdeki (included) potansiyel hatalı girişleri filtreler.
print("=== 1) AYNI RECORD_ID, FARKLI TITLE ===")
same_id_diff_title = included.groupby('record_id').filter(lambda x: len(x['title'].unique()) > 1)
print(f"Toplam {len(same_id_diff_title)} kayıt bulundu.")
display(same_id_diff_title)
print("\n" + "="*80 + "\n")

# --- 2. Adım: Farklı başlıkların detaylı analizi ---
# Bu adım, yukarıda bulunan kayıtları özetleyerek daha anlaşılır hale getirir.
print("=== 2) FARKLI BAŞLIKLARIN DETAYLI ANALİZİ ===")

# 'record_id'ye göre grupla ve her bir grup için 'title' listesini oluştur.
grouped_titles = same_id_diff_title.groupby('record_id').agg(
    titles_list=('title', list),
    authors_list=('authors', list)
).reset_index()

# Her bir 'record_id' için kaç farklı başlık olduğunu gösteren yeni bir sütun ekle.
grouped_titles['title_count'] = grouped_titles['titles_list'].apply(len)

# Analiz tablosunu göster.
display(grouped_titles)
```

```python
import pandas as pd
from IPython.display import display

print("=== 1) AYNI RECORD_ID, AYNI TITLE - HAM VERİ ===")
# 'record_id' ve 'title' sütunlarını kullanarak tam kopyaları bulur.
# 'keep=False' parametresi, tüm kopyaları (hem ilk hem de sonraki tekrar edenleri) gösterir.
same_id_same_title = included[included.duplicated(subset=['record_id', 'title'], keep=False)]
print(f"Toplam {len(same_id_same_title)} kayıt bulundu.")
display(same_id_same_title)
print("\n" + "="*80 + "\n")
```

```python
print("=== 2) DUPLİKE KAYITLARIN ANALİZİ ===")

# Bulunan kopyaları 'record_id' ve 'title'a göre gruplayıp analiz eder.
grouped_duplicates = same_id_same_title.groupby(['record_id', 'title']).agg(
    authors_list=('authors', list),
    count=('record_id', 'count')
).reset_index()

# Sadece 2'den fazla tekrar edenleri göstermek için filtreleme yapabilirsiniz (isteğe bağlı).
# grouped_duplicates = grouped_duplicates[grouped_duplicates['count'] > 1]

# Analiz tablosunu gösterir.
display(grouped_duplicates)
```

```python
import pandas as pd
from IPython.display import display

print("=== 1) FARKLI RECORD_ID, AYNI TITLE - HAM VERİ ===")
# 'title' sütununa göre grupla ve 'record_id'lerin benzersiz sayısının 1'den fazla olduğu grupları filtrele.
diff_id_same_title = included.groupby('title').filter(lambda x: len(x['record_id'].unique()) > 1)
print(f"Toplam {len(diff_id_same_title)} kayıt bulundu.")
display(diff_id_same_title)
print("\n" + "="*80 + "\n")
```

```python
print("=== 2) DUPLİKE BAŞLIKLARIN ANALİZİ ===")

# Bulunan yinelenen başlıkları 'title'a göre gruplayıp analiz eder.
grouped_duplicates = diff_id_same_title.groupby('title').agg(
    record_ids=('record_id', list),     # Tüm 'record_id'leri liste olarak topla.
    authors_list=('authors', list)        # Tüm 'authors'ları liste olarak topla.
).reset_index()

# Her bir başlığın kaç kez tekrar ettiğini gösteren yeni bir sütun ekle.
grouped_duplicates['repeat_count'] = grouped_duplicates['record_ids'].apply(len)

# Analiz tablosunu gösterir.
display(grouped_duplicates)
```

```python
import pandas as pd
from IPython.display import display

# --- 1. Adım: Farklı ID'lere sahip, aynı başlıklı kayıtları bulma ---
print("=== 1) FARKLI RECORD_ID, AYNI TITLE ===")
diff_id_same_title = screening.groupby('title').filter(lambda x: len(x['record_id'].unique()) > 1)
print(f"Toplam {len(diff_id_same_title)} kayıt bulundu.")
display(diff_id_same_title)
print("\n" + "="*80 + "\n")

# --- 2. Adım: Yinelenen başlıkları detaylı olarak gruplama ve analiz etme ---
print("=== 2) DUPLIKE BAŞLIKLARIN DETAYLI ANALİZİ ===")

grouped_duplicates = diff_id_same_title.groupby('title').agg(
    record_ids=('record_id', list),
    authors_list=('authors', list)
).reset_index()

grouped_duplicates['repeat_count'] = grouped_duplicates['record_ids'].apply(len)

display(grouped_duplicates)

# --- 3. Adım: Analiz tablosunu CSV dosyasına kaydetme ---
# Oluşturulan `grouped_duplicates` DataFrame'ini bir CSV dosyasına kaydeder.
# 'index=False' parametresi, Pandas indeks numarasının dosyaya yazılmasını engeller.
output_file = 'duplicate_titles_analysis.csv'
grouped_duplicates.to_csv(output_file, index=False)

print(f"\nAnaliz sonuçları '{output_file}' dosyasına başarıyla kaydedildi.");
```

```python
# Farklı ID'lere sahip, aynı başlıklı kayıtları bul
diff_id_same_title = screening.groupby('title').filter(lambda x: len(x['record_id'].unique()) > 1)

# Bu kayıtların başlıklarını ve bunlara ait ID'leri bir sözlükte grupla
grouped_titles = diff_id_same_title.groupby('title')['record_id'].apply(list).to_dict()

# Düzeltme için bir kural belirle: Her grup için ilk record_id'yi doğru kabul et
# Diğer tüm record_id'leri bu ilk id ile değiştireceğiz
correction_map = {}
for title, ids in grouped_titles.items():
    primary_id = ids[0]
    for secondary_id in ids[1:]:
        correction_map[secondary_id] = primary_id
```

```python
# screening DataFrame'indeki tüm record_id'leri correction_map'e göre güncelle
screening['record_id'] = screening['record_id'].replace(correction_map)

print("screenign veri setindeki 'record_id' duplikasyonları başarıyla düzeltildi.")
```

```python
# Düzeltme sonrası 'farklı ID, aynı title' durumunu tekrar kontrol et
print("\n=== DÜZELTME SONRASI KONTROL ===")
diff_id_same_title_after = screening.groupby('title').filter(lambda x: len(x['record_id'].unique()) > 1)

if diff_id_same_title_after.empty:
    print("Mükemmel! Artık 'farklı record_id, aynı title' problemi bulunmuyor.")
    # Bu aşamada, aynı ID, aynı başlık olan kayıtlar oluştuğu için
    # onları da temizlememiz gerekecek
    same_id_same_title = screening[screening.duplicated(subset=['record_id', 'title'], keep='first')]
    print(f"Düzeltme sonrası oluşan tam duplike kayıt sayısı: {len(same_id_same_title)}")
    
    # Tam duplikeleri kaldır
    screening.drop_duplicates(subset=['record_id', 'title'], keep='first', inplace=True)
    
    print("Tam duplike kayıtlar temizlendi. screening veri seti şimdi daha temiz.")
else:
    print("Hata! Bazı kayıtlar düzeltilemedi.")
    display(diff_id_same_title_after)
```

```python
import pandas as pd

# --- Adım 1: screening_log.csv dosyasını yükle ---
# Dosya yolunun doğru olduğundan emin ol
try:
    screening_log = pd.read_csv('data/screening_log.csv')
    print("screening_log.csv başarıyla yüklendi.")
    display(screening_log.head())
except FileNotFoundError:
    print("HATA: 'screening_log.csv' dosyası bulunamadı. Lütfen dosya yolunu kontrol et.")
    # Eğer dosya yoksa, aşağıdaki kod çalışmayacağı için işlemi burada durdurabiliriz.
    screening_log = None

# --- Adım 2: Çapraz Kontrolü Gerçekleştir ---
if screening_log is not None:
    # Günlük dosyasından "dahil edildi" olarak işaretlenenlerin ID'lerini al
    # !!! DİKKAT: 'status' ve 'included' değerlerini kendi verinize göre güncelleyin !!!
    try:
        supposed_to_be_included_ids = set(screening_log[screening_log['decision'] == 'Include']['record_id'])
        print(f"\nLog dosyasına göre dahil edilmesi gereken {len(supposed_to_be_included_ids)} adet kayıt var.")
    
        # Mevcut 'included' DataFrame'indeki ID'leri al
        # Bu DataFrame'in daha önceki adımlarda temizlendiğini varsayıyoruz.
        actually_included_ids = set(included['record_id'])
        print(f"Mevcut 'included' DataFrame'inde ise {len(actually_included_ids)} adet kayıt var.")
    
        print("\n" + "="*80 + "\n")
        print("Çapraz kontrol sonuçları:")
    
        # --- Kontrol 1: Log'da 'included' ama DataFrame'de YOK ---
        # Bunlar, dahil edilmesi gerekirken kaybolmuş olabilecek kayıtlardır.
        missing_from_df = supposed_to_be_included_ids - actually_included_ids
        if missing_from_df:
            print(f"\n[!] UYARI: Log'da 'included' olarak işaretli ama DataFrame'de bulunamayan {len(missing_from_df)} kayıt tespit edildi:")
            print(list(missing_from_df))
        else:
            print("\n[✓] Harika! Log'da 'included' olan tüm kayıtlar DataFrame'de mevcut.")
    
        # --- Kontrol 2: DataFrame'de VAR ama Log'da 'included' DEĞİL ---
        # Bunlar, yanlışlıkla dahil edilmiş olabilecek kayıtlardır.
        extra_in_df = actually_included_ids - supposed_to_be_included_ids
        if extra_in_df:
            print(f"\n[!] UYARI: DataFrame'de mevcut ama Log'da 'included' olarak işaretlenmemiş {len(extra_in_df)} kayıt tespit edildi:")
            print(list(extra_in_df))
        else:
            print("[✓] Harika! DataFrame'deki tüm kayıtlar Log dosyasında 'included' olarak doğrulanmıştır.")

    except KeyError as e:
        print(f"\n[HATA]: Kodda belirtilen bir sütun adı bulunamadı: {e}")
        print("Lütfen `screening_log` DataFrame'indeki 'status' ve 'record_id' gibi sütun adlarını kontrol edin.")
```

```python
import pandas as pd
import os

# --- Adım 1: Dosya Yollarını Tanımlama ---
# BURAYI KENDİ BİLGİSAYARINIZDAKİ TAM YOL İLE DEĞİŞTİRİN!
# ÖRNEK: elicit_file_path = "C:\\Users\\KullanıcıAdı\\Downloads\\elicit_with_record_ids-2.csv"
elicit_file_path = os.path.join(os.path.expanduser('~'), 'Downloads', 'elicit_with_record_ids-2.csv')
studies_file_path = os.path.join('data', 'included_studies.csv')

print("İşlem başlıyor...")
print(f"Ana veri dosyası: '{elicit_file_path}'")

try:
    # --- Adım 2: Ana CSV Dosyasını DataFrame'e Yükleme ---
    df_elicit = pd.read_csv(elicit_file_path)
    print(f"\n✅ '{elicit_file_path}' dosyası başarıyla yüklendi.")
    print(f"Başlangıçtaki toplam satır sayısı: {df_elicit.shape[0]}")

    # --- Adım 3: Veri Tutarsızlıklarını Tespit Etme ve Ayırma ---
    print("\n" + "="*50)
    print("YENİ ADIM: Tutarsız Verileri Tespit Et ve Ayrı Bir DataFrame'e Topla")
    print("="*50)

    # KONTROL 1: Farklı record_id, Aynı title
    #----------------------------------------------------
    print("\n>>> KONTROL 1: Aynı başlığa sahip FARKLI record_id'ler aranıyor...")
    title_id_counts = df_elicit.groupby('Title')['record_id'].nunique()
    conflicting_titles = title_id_counts[title_id_counts > 1].index
    
    # KONTROL 2: Aynı record_id, Farklı title
    #----------------------------------------------------
    print(">>> KONTROL 2: Aynı record_id'ye sahip FARKLI başlıklar aranıyor...")
    id_title_counts = df_elicit.groupby('record_id')['Title'].nunique()
    conflicting_ids = id_title_counts[id_title_counts > 1].index
    
    # Tespit edilen tüm tutarsız satırları filtreleyerek alalım
    inconsistent_type1_df = df_elicit[df_elicit['Title'].isin(conflicting_titles)]
    inconsistent_type2_df = df_elicit[df_elicit['record_id'].isin(conflicting_ids)]

    if not inconsistent_type1_df.empty or not inconsistent_type2_df.empty:
        # İki türdeki tutarsız DataFrame'leri birleştirelim
        df_all_inconsistent = pd.concat([inconsistent_type1_df, inconsistent_type2_df]).drop_duplicates().reset_index(drop=True)
        
        print(f"\n🚨 TOPLAM {df_all_inconsistent.shape[0]} ADET TUTARSIZ SATIR BULUNDU.")
        print("Bu satırlar ayrı bir DataFrame'de toplanıyor ve ana veriden çıkarılacak.")
        print("\nİşte incelenmesi gereken tutarsız verilerden ilk 5 örnek:")
        display(df_all_inconsistent.head())

        # YENİ TEMİZ DATAFRAME: Tutarsız satırların indekslerini kullanarak ana DataFrame'den düşelim
        clean_indices = df_elicit.index.difference(df_all_inconsistent.index)
        df_elicit_clean = df_elicit.loc[clean_indices]
        
        print(f"\nOrijinal satır sayısı: {df_elicit.shape[0]}")
        print(f"Ayrılan tutarsız satır sayısı: {df_all_inconsistent.shape[0]}")
        print(f"Analize devam edilecek TEMİZ satır sayısı: {df_elicit_clean.shape[0]}")
    else:
        print("\n✅ GÜZEL HABER: Veri setinde herhangi bir tutarsızlık bulunamadı.")
        df_all_inconsistent = pd.DataFrame()
        df_elicit_clean = df_elicit.copy()

    # --- Adım 4: Filtreleme İçin Kullanılacak DataFrame'i Yükleme ---
    print("\n" + "="*50)
    print("Filtreleme Aşamasına Geçiliyor...")
    print("="*50)
    
    df_studies = pd.read_csv(studies_file_path)
    print(f"\n✅ '{studies_file_path}' dosyası başarıyla yüklendi.")
    print(f"Filtreleme için {df_studies.shape[0]} adet benzersiz ID bulundu.")

    # --- Adım 5: Saklanacak 'record_id'leri Bir Kümeye (Set) Alma ---
    record_ids_to_keep = set(df_studies['record_id'].unique())
    
    # --- Adım 6: TEMİZ DataFrame'i Filtreleme ---
    df_final_filtered = df_elicit_clean[df_elicit_clean['record_id'].isin(record_ids_to_keep)]
    
    print(f"\n✅ ID'lere göre filtreleme işlemi TEMİZ VERİ üzerinde tamamlandı.")
    print(f"Son filtreleme sonrası kalan satır sayısı: {df_final_filtered.shape[0]}")

    # --- Adım 7: Sonuçları Gözden Geçirme ---
    print("\nNihai olarak filtrelenmiş ve temizlenmiş verinin ilk 5 satırı (`df_final_filtered`):")
    display(df_final_filtered.head())
    
except FileNotFoundError as e:
    print(f"\n❌ HATA: Dosya bulunamadı! Lütfen dosya yollarını kontrol edin.")
    print(f"Hata detayı: {e}")
except KeyError as e:
    print(f"\n❌ HATA: Sütun bulunamadı! Dosyalarda 'record_id' veya 'title' adında bir sütun olduğundan emin olun.")
    print(f"Hata detayı: {e}")
except Exception as e:
    print(f"\nBeklenmedik bir hata oluştu: {e}")
```

```python
import pandas as pd
import os

# --- Adım 1: Dosya Yollarını ve Veri Setini Tanımlama ---
# Veri setinizin dosya yolu
dataframe_file_path = "/data/final_filtered_data.csv"

# Arama yapılacak ana klasörün yolu
search_directory = "OneDrive - ASKERİ FABRİKA VE TERSANE İŞLETME A.Ş (ASFAT)/kisiselAlan/AKU_Doktora"

# --- Adım 2: Veri Setini Yükle ve Filtrele ---
try:
    # DataFrame'i yükle
    df = pd.read_csv(dataframe_file_path)

    # '-' ve boş değerler hariç Filename sütunundaki benzersiz dosya isimlerini al
    # Boş string ('') ve NaN (eksik veri) değerlerini de kontrol ediyoruz
    valid_filenames = df[df['Filename'].notna() & (df['Filename'] != '-') & (df['Filename'] != '')]['Filename'].unique()
    
    print(f"Toplam {len(valid_filenames)} adet dosya adı aranacak.\n")
    
except FileNotFoundError:
    print(f"Hata: Veri seti '{dataframe_file_path}' konumunda bulunamadı. Lütfen dosya yolunu kontrol edin.")
    exit()

# --- Adım 3: Belirtilen Klasörde Dosya Arama ---
found_files = []

# os.walk() fonksiyonu ile ana dizini ve tüm alt dizinlerini dolaş
print(f"'{search_directory}' klasöründe dosya araması başlatılıyor...")
for root, dirs, files in os.walk(search_directory):
    for file in files:
        if file in valid_filenames:
            # Eğer dosya adı aranacaklar listesinde varsa, tam yolunu al
            full_path = os.path.join(root, file)
            found_files.append({'filename': file, 'full_path': full_path})
            print(f"  Bulundu: {full_path}")

# --- Adım 4: Bulunan Dosyaları DataFrame'e Dönüştürme ve Eşleştirme ---
if not found_files:
    print("\nBelirtilen dosya adlarından hiçbiri bulunamadı.")
else:
    found_df = pd.DataFrame(found_files)
    
    # Ana DataFrame ile bulunan dosyaları içeren DataFrame'i birleştir
    # 'Filename' sütununu referans alarak birleştirme yapıyoruz
    merged_df = pd.merge(df, found_df, on='Filename', how='inner')
    
    # Yalnızca gerekli sütunları (record_id, Filename, full_path) seç
    final_df = merged_df[['record_id', 'Filename', 'full_path']]
    
    print(f"\n✅ Toplam {len(final_df)} adet kayıt başarıyla eşleştirildi.")

    # --- Adım 5: Sonuçları CSV Olarak Kaydetme ---
    output_filename = 'matched_files_with_record_ids.csv'
    final_df.to_csv(output_filename, index=False, encoding='utf-8-sig')
    
    print(f"\nSonuçlar '{output_filename}' dosyasına kaydedildi.")
    print("İşte ilk 5 sonuç:")
    display(final_df.head())
```

```python
import pandas as pd
import os

# --- Adım 1: Dosya Yollarını ve Veri Setini Tanımlama ---
# Veri setinizin dosya yolu
# Bu yolun projenizin ana dizinine göreceli olduğunu varsayıyoruz.
dataframe_file_path = os.path.join('data', 'final_filtered_data.csv')

# Arama yapılacak ana klasörün yolu
search_directory = "OneDrive - ASKERİ FABRİKA VE TERSANE İŞLETME A.Ş (ASFAT)/kisiselAlan/AKU_Doktora"

# Arama dışında tutulacak klasörün adı
excluded_folder_name = 'gitReview'

print("İşlem başlatılıyor...")
print(f"Veri seti yolu: '{dataframe_file_path}'")
print(f"Arama yapılacak ana dizin: '{search_directory}'")
print(f"Arama dışında tutulacak klasör: '{excluded_folder_name}'")

# --- Adım 2: Veri Setini Yükle ve Filtrele ---
try:
    df = pd.read_csv(dataframe_file_path)
    valid_filenames = df[df['Filename'].notna() & (df['Filename'] != '-') & (df['Filename'] != '')]['Filename'].unique()
    print(f"\nToplam {len(valid_filenames)} adet dosya adı aranacak.\n")

except FileNotFoundError:
    print(f"Hata: Veri seti '{dataframe_file_path}' konumunda bulunamadı. Lütfen dosya yolunu kontrol edin.")
    exit()
except KeyError:
    print("Hata: 'Filename' sütunu bulunamadı. Lütfen sütun adını kontrol edin.")
    exit()

# --- Adım 3: Belirtilen Klasörde Dosya Arama ---
found_files = []

# Dosya yolu mevcut değilse uyarı ver
if not os.path.exists(search_directory):
    print(f"Hata: Arama yapılacak dizin '{search_directory}' bulunamadı. Lütfen dizin yolunu kontrol edin.")
    exit()

print(f"'{search_directory}' klasöründe dosya araması başlatılıyor...")

for root, dirs, files in os.walk(search_directory):
    # Arama dışında tutulacak klasörü kontrol et
    # Bu kontrol, alt klasörlere inmeden önce yapılır
    if excluded_folder_name in root.split(os.sep):
        print(f"  '{root}' dizini arama dışında tutuluyor.")
        continue # Bu dizini atla ve bir sonraki dizine geç

    for file in files:
        if file in valid_filenames:
            full_path = os.path.join(root, file)
            found_files.append({'filename': file, 'full_path': full_path})
            print(f"  Bulundu: {full_path}")

# --- Adım 4: Bulunan Dosyaları DataFrame'e Dönüştürme ve Eşleştirme ---
if not found_files:
    print("\nBelirtilen dosya adlarından hiçbiri bulunamadı.")
else:
    found_df = pd.DataFrame(found_files)
    merged_df = pd.merge(df, found_df, on='Filename', how='inner')
    final_df = merged_df[['record_id', 'Filename', 'full_path']]

    print(f"\n✅ Toplam {len(final_df)} adet kayıt başarıyla eşleştirildi.")

    # --- Adım 5: Sonuçları CSV Olarak Kaydetme ---
    output_filename = 'matched_files_with_record_ids.csv'
    final_df.to_csv(output_filename, index=False, encoding='utf-8-sig')

    print(f"\nSonuçlar '{output_filename}' dosyasına kaydedildi.")
    print("İşte ilk 5 sonuç:")
    display(final_df.head())
```
